```python
import sys, os, csv
sys.path.insert(0, "/content")
import numpy as np, xccnet
from PIL import Image
records = xccnet.collect_kermany_records(xccnet.resolve_kermany_root())
test = xccnet.stratified_split(records, seed=42)["test"]
os.makedirs("/content/results", exist_ok=True)
with open("/content/results/mask_coverage.csv", "w", newline="") as h:
    w = csv.writer(h); w.writerow(["file", "label", "mask_fraction"])
    for src, label in test:
        p = "/content/work/cache_clahe_lung256/" + xccnet._cache_name(src, label)
        a = np.asarray(Image.open(p))
        w.writerow([os.path.basename(p), label, f"{(a > 0).mean():.4f}"])
print("wrote", len(test))
```
